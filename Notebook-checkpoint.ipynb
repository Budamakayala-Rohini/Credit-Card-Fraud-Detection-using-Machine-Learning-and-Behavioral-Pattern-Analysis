```python
ML_Model = []
accuracy = []
precision = []
recall = []
f1score = []

def storeResults(model, a, b, c, d):
    ML_Model.append(model)
    accuracy.append(round(a, 4))
    precision.append(round(b, 4))
    recall.append(round(c, 4))
    f1score.append(round(d, 4))
```

```python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, classification_report, confusion_matrix,
    ConfusionMatrixDisplay, roc_curve, auc
)
from sklearn.preprocessing import label_binarize


def evaluate_model(y_test, y_pred, y_proba, labels, history=None, model_name="Model"):
    # Metrics (macro averaging for multiclass)
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, average="macro")
    rec = recall_score(y_test, y_pred, average="macro")
    f1 = f1_score(y_test, y_pred, average="macro")

    print(f"\nClassification Report - {model_name}")
    print(classification_report(y_test, y_pred, target_names=labels))

    fig, axes = plt.subplots(1, 2, figsize=(15, 6))

    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels)
    disp.plot(cmap='viridis', xticks_rotation=90, ax=axes[0], colorbar=False)
    axes[0].set_title(f"Confusion Matrix - {model_name}")

    # ROC Curve (for multiclass)
    if len(np.unique(y_test)) > 2:
        y_test_bin = label_binarize(y_test, classes=range(len(labels)))
        fpr, tpr, roc_auc = {}, {}, {}

        for i in range(len(labels)):
            fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], y_proba[:, i])
            roc_auc[i] = auc(fpr[i], tpr[i])
            axes[1].plot(fpr[i], tpr[i], lw=2, label=f"{labels[i]} (AUC={roc_auc[i]:.2f})")

        axes[1].plot([0, 1], [0, 1], 'k--', lw=2)
        axes[1].set_title(f"ROC Curve - {model_name}")
        axes[1].legend(loc="lower right", fontsize=8)

    else:  # binary case
        # take probability of positive class (index 1)
        if y_proba.ndim > 1:
            y_score = y_proba[:, 1]
        else:
            y_score = y_proba

        fpr, tpr, _ = roc_curve(y_test, y_score)
        roc_auc_val = auc(fpr, tpr)

        axes[1].plot(fpr, tpr, lw=2, label=f"AUC = {roc_auc_val:.2f}")
        axes[1].plot([0, 1], [0, 1], 'k--', lw=2)
        axes[1].legend(loc="lower right", fontsize=8)

    plt.tight_layout()
    plt.show()

    # Training curves
    if history:
        fig, axes = plt.subplots(1, 2, figsize=(15, 6))

        if isinstance(history, dict):
            history_dict = history
        else:
            history_dict = history.history

        axes[0].plot(history_dict["accuracy"], label="Train Accuracy")
        axes[0].plot(history_dict["val_accuracy"], label="Val Accuracy")
        axes[0].set_title(f"Training Accuracy - {model_name}")
        axes[0].set_xlabel("Epochs")
        axes[0].set_ylabel("Accuracy")
        axes[0].legend()

        axes[1].plot(history_dict["loss"], label="Train Loss")
        axes[1].plot(history_dict["val_loss"], label="Val Loss")
        axes[1].set_title(f"Training Loss - {model_name}")
        axes[1].set_xlabel("Epochs")
        axes[1].set_ylabel("Loss")
        axes[1].legend()

        plt.tight_layout()
        plt.show()

    return acc, prec, rec, f1
```

# Deep Learning

```python
from tensorflow.keras import utils
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, callbacks
from tensorflow.keras.models import load_model


n_features = X_train.shape[1]

# Convert DataFrame -> numpy array and reshape for CNN input
X_train_dl = np.array(X_train).reshape((-1, n_features, 1))
X_test_dl  = np.array(X_test).reshape((-1, n_features, 1))

# One-hot encode target
num_classes = len(np.unique(y_train))
y_train_cat = utils.to_categorical(y_train, num_classes)
y_test_cat = utils.to_categorical(y_test, num_classes)
```

```python
callbacks_list = [
    callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
]

input_shape = (X_train.shape[1],)
```

# LSTM

```python
input_dim = X_train.shape[1]
lstm_model = models.Sequential([
    layers.Input(shape=(input_dim, 1)),
    layers.LSTM(128, return_sequences=True),
    layers.Dropout(0.3),
    layers.LSTM(64),
    layers.Dropout(0.2),
    layers.Dense(128, activation='relu'),
    layers.Dense(num_classes, activation='softmax')
])
lstm_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
lstm_model.summary()
```

```python
if os.path.exists("Models/lstm_model.h5") and os.path.exists("Models/lstm_history.sav"):
    print("Loading existing LSTM model...")
    lstm_model = load_model("Models/lstm_model.h5")
    history_lstm_dict = joblib.load("Models/lstm_history.sav")
else:
    print("Training new LSTM model...")
    history_lstm = lstm_model.fit(
        X_train_dl, y_train_cat,
        epochs=5,
        batch_size=256,
        validation_split=0.2,
        callbacks=callbacks_list,
        verbose=1
    )

    # Save model + history
    lstm_model.save("Models/lstm_model.h5")
    joblib.dump(history_lstm.history, "Models/lstm_history.sav")
    history_lstm_dict = history_lstm.history
```

```python
y_proba_lstm = lstm_model.predict(X_test_dl) 
y_pred_lstm = np.argmax(y_proba_lstm, axis=1)

acc_lstm, prec_lstm, rec_lstm, f1_lstm = evaluate_model(
    y_test, y_pred_lstm, y_proba_lstm, labels, history=history_lstm_dict, model_name="LSTM"
)
```

```python
storeResults("LSTM", acc_lstm, prec_lstm, rec_lstm, f1_lstm)
```

```python
if os.path.exists("Models/gru_model.h5") and os.path.exists("Models/gru_history.sav"):
    print("Loading existing GRU model...")
    gru_model = load_model("Models/gru_model.h5")
    history_gru_dict = joblib.load("Models/gru_history.sav")
else:
    print("Training new GRU model...")
    history_gru = gru_model.fit(
        X_train_dl, y_train_cat,
        epochs=5,
        batch_size=256,
        validation_split=0.2,
        callbacks=callbacks_list,
        verbose=1
    )

    # Save model and history
    gru_model.save("Models/gru_model.h5")
    joblib.dump(history_gru.history, "Models/gru_history.sav")
    history_gru_dict = history_gru.history
```

```python
y_proba_gru = gru_model.predict(X_test_dl)
y_pred_gru = np.argmax(y_proba_gru, axis=1)

acc_gru, prec_gru, rec_gru, f1_gru = evaluate_model(
    y_test, y_pred_gru, y_proba_gru, labels, history=history_gru_dict, model_name="GRU"
)
```

```python
storeResults("GRU", acc_gru, prec_gru, rec_gru, f1_gru)
```

# Hybrid (LSTM + GRU + MLP) Model

```python
input_dim = X_train.shape[1]

X_train_hybrid = np.array(X_train).reshape((-1, input_dim, 1))
X_test_hybrid  = np.array(X_test).reshape((-1, input_dim, 1))

hybrid_model = models.Sequential([
    layers.Input(shape=(input_dim, 1)),
    
    # LSTM branch
    layers.LSTM(128, return_sequences=True),
    layers.Dropout(0.3),
    
    # GRU branch
    layers.GRU(64),
    layers.Dropout(0.3),
    
    # MLP block
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(64, activation='relu'),
    layers.Dense(num_classes, activation='softmax')
])

hybrid_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
hybrid_model.summary()
```

```python
if os.path.exists("Models/hybrid_lstm_gru_mlp_model.h5") and os.path.exists("Models/hybrid_history.sav"):
    print("Loading existing Hybrid LSTM+GRU+MLP model...")
    hybrid_model = load_model("Models/hybrid_lstm_gru_mlp_model.h5")
    history_hybrid_dict = joblib.load("Models/hybrid_history.sav")
else:
    print("Training new Hybrid LSTM+GRU+MLP model...")
    history_hybrid = hybrid_model.fit(
        X_train_hybrid, y_train_cat,
        epochs=5,
        batch_size=256,
        validation_split=0.2,
        callbacks=callbacks_list,
        verbose=1
    )

    hybrid_model.save("Models/hybrid_lstm_gru_mlp_model.h5")
    joblib.dump(history_hybrid.history, "Models/hybrid_history.sav")
    history_hybrid_dict = history_hybrid.history
```

```python
y_proba_hybrid = hybrid_model.predict(X_test_hybrid)
y_pred_hybrid = np.argmax(y_proba_hybrid, axis=1)

acc_hybrid, prec_hybrid, rec_hybrid, f1_hybrid = evaluate_model(
    y_test, y_pred_hybrid, y_proba_hybrid, labels, history=history_hybrid_dict, model_name="Hybrid_LSTM_GRU_MLP"
)
```

```python
storeResults("Hybrid_LSTM_GRU_MLP", acc_hybrid, prec_hybrid, rec_hybrid, f1_hybrid)
```
